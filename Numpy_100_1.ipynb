# 56. Generate a generic 2D Gaussian-like array (★★☆)
# `hint: np.meshgrid, np.exp`

````python

#Gpt help

n = 10
x = np.linspace(-1, 1, n)
y = np.linspace(-1, 1, n)


X, Y = np.meshgrid(x, y)

sigma = 0.5

G = np.exp(-(X**2 + Y**2) / (2 * sigma**2))

print(G)
````
