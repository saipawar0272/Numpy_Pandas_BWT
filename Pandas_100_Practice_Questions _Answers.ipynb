### Q56. Detect outliers in a column.
> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.

---

```python
print(data)

Q1 = data['age'].quantile(0.25)
print('Q1 :',Q1)

Q3 = data['age'].quantile(0.75)
print('Q3 :',Q3)

IQR = Q3 - Q1
print('IQR : ', IQR)

below = Q1 - 1.5*IQR
above = Q3 + 1.5*IQR

print('below :',below)
print('above :', above)

```

### Q59. Handle mixed data types in a column.
> 💡 **Hint:** Use `pd.to_numeric(df["col"], errors="coerce")` to convert to numbers, turning invalid strings into `NaN`.

---

```python
#print(data)


result = {
    "id": [101, 102, "103A", 104, None, 106],
    "name": ["John", "Alice", None, "Emma", 123, "David"],
    "age": [25, "30", None, 35, "Unknown", 40],
    "salary": [50000, "60000", None, 75000.50, "N/A", 80000],
    "bonus": [5000, None, "5500", 7000, "Not Given", 8000],
    "married": ["Yes", "No", True, False, None, "Yes"],
    "department": ["IT", "HR", 101, None, "Sales", "Finance"]
}

result = pd.DataFrame(result)
#print(result)

#result['columns'] = (result.columns.str.upper())

result.columns = (result.columns.str.upper())
#print(result)



result["BONUS"] = pd.to_numeric(result["BONUS"], errors='coerce')
#print(result)

result['AGE'] = pd.to_numeric(result['AGE'], errors='coerce')

#result['ID', 'SALARY'] = pd.to_numeric(result['ID', 'SALARY'], errors='coerce')  #wrong way

result[['ID','SALARY']] = result[['ID', 'SALARY']].apply(
    pd.to_numeric, errors = 'coerce'
)


result['MARRIED'] = result['MARRIED'].astype('category')

result['SALARY'] = result['SALARY'].mean()

mean_bonus = result['BONUS'].mean()
print('mean_bonus',mean_bonus)
result['BONUS'] = result['BONUS'].fillna(mean_bonus).mean()

mean_age = result['AGE'].mean()
print('mean_age', mean_age)

result['AGE'] = result['AGE'].fillna(mean_age)

print(result)
#print(result.dtypes)
```
